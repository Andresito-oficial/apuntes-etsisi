# Parte I: Colecciones
Las colecciones son elementos iterables con múltiples elementos, recogiendo strings y números. Una peculiaridad de Python es que en las colecciones se pueden mezclar tipos de dato. Da igual el tipo en ese contexto, una lista, tupla y conjunto pueden tener varios elementos de diversos tipos de datos ya sean integer, float o string. Adicionalmente los conjuntos tienen funciones asociadas que se pueden usar para expandir, ampliar etc.

Vamos a ver qué diferencia a cada colección.

## Conjuntos
Los conjuntos se componen de varios elementos únicos, por ejemplo {1 2 3} o {"a" "b" "c"}. Son limitados y no se puede acceder a ellos puesto que se considera que no tienen un orden. Sin embargo se pueden modificar, expandiendo y reduciendo.

```python
operators = {'a', 'b', 'c','d'}

for elem in operators: # No saldrán ordenados!
  print(elem)
```

Se puede ver si un elemento pertenece a cualquier colección mediante *in*

```python
print('a' in operators)
print({'a','b'} in operators) # No funciona
```

Podemos incluir elementos en un conjunto usando la operacion _.update()_ para colecciones y _.add()_ para elementos.

```python
operators = set() # Para crear un conjunto vacio
z = 55
operators.add(z)
print(operators)
operators.update('a', 'b', {'e', 'f'})
print(operators)
```

Las listas, al contrario que los conjuntos sí permiten acceso desde la indización, pueden ordenarse, extraer subconjuntos etc. Además se pueden repetir elementos en su interior al contrario que con los conjuntos. Las listas además pueden ser accedidas mediante *slicing*, una extracción de una sublista a partir de un índice inicial y otro final. También probamos otros operadores que tienen comportamientos útiles y únicos a las listas de Python. Vemos algunos ejemplos:

## Args y KWargs, argumentos sin especificar
Las palabras clave args y kwargs representan una lista y un diccionario respectivamente, en el que se almacenan los valores que no hayan sido especificados en la entrada. Vemos ejemplos:

```python
def sumatorio(*args):
  suma = 0
  for item in args:
    suma+=item
  return suma

print(sumatorio(1, 2, 3, 4))
```

```python
def operacion(**kwargs):
  s = 0
  if 'a' in kwargs.keys():
    s += kwargs['a']
  if 'b' in kwargs.keys():
    s -= kwargs['b']
  if 'c' in kwargs.keys():
    s *= kwargs['c']
  return s

print(operacion(a=2, c=10, d=1)) # ignora d=1!
```

## La función lambda
La función lambda es un tipo de función de usar y tirar, se declara en el instante y no se utiliza nada más que para una operación. # Parte III: Comprehensiones
Una de las capacidades únicas y más potentes de python son las comprehensiones. Una comprehensión es una forma de abreviar definiciones de listas y diccionarios mediante la semántica de un bucle. También pueden lanzarse como argumento de algunas funciones. Vamos a ver tres ejemplos, una como lista, otra como diccionario y una última como generador.

## Ejemplo de lista
Supongamos que queremos coger cada caracter de un string y ponerle un número delante, para ello vamos a utilizar una comprensión de lista y la función integrada **enumerate**. Esta función devuelve el índice de cada elemento de la colección que recibe por entrada en formato tupla (indice, elemento), esta tupla es volcada sobre las variables *i* y *c* automáticamente.

```python
listcomp = [str(i) + c for i, c in enumerate('holamundo!')]
print(listcomp)
```

También podemos usar las propias listas o *range* como iteradores:

```python
A = [1, 2, 3, 4, 5]
A2 = [2*i for i in A]
print(A2)
```

```python
squares = [x**2 for x in range(1, 11)]
print(squares)
evens = [x for x in range(21) if x % 2 == 0]
print(evens)
```

## Ejemplo de generador
Esta vez vamos a contar con dos listas y queremos hallar el mínimo de la resta entre la lista A y la lista B. Podemos introducir una comprehensión por entrada a la función **max** que computará el máximo de los valores. Esta vez utilizaremos la función **zip** para emparejar ambas listas y multiplicar sus valores.

```python
A = [1, 2, 3]
B = [4, 3, -1]
prodsum = min(a-b for a,b in zip(A, B))
print(prodsum)
```

## Copy y deepcopy

El módulo copy sirve para copiar listas. Copy crea una copia "shallow" que no copia referencias internas, mientras que deepcopy hace una copia "profunda" recursivamente copiando los elementos del objeto. Vemos ejemplos:

```python
from copy import copy
lista = list(range(5))
nueva_lista_falsa = lista
nueva_lista_copia = copy(lista)
lista[0] = 100
print(nueva_lista_falsa[0])
print(nueva_lista_copia[0])
```

```python
from copy import copy, deepcopy
lista = [0,10, [10,10,10]]
nueva_lista_copia_shallow = copy(lista)
nueva_lista_copia_deep = deepcopy(lista)

lista[2][0] = 100 # La lista interna no es copiada por copy!
print(nueva_lista_copia_shallow[2][0])
print(nueva_lista_copia_deep[2][0])
```

## NumPy y los tensores
La librería numpy ofrece muchas opciones para hacer cálculos matemáticos pero lo principal que nos trae es la capacidad de operar con tensores, multiplicaciones, número aleatorios y mucho más. Un tensor es parecido a una matriz, con la diferencia que puede tener más de dos dimensiones.

```python
import numpy as np

a = np.random.rand(5, 10) # matriz de 5x10 generada aleatoriamente
b = np.random.rand(5, 10) # matriz de 5x10 generada aleatoriamente
c = np.random.rand(1, 10) # matriz de 1x10 generada aleatoriamente
d = np.random.rand(10, 2) # matriz de 10x2 generada aleatoriamente

print('a')
print(a)
print('\na+1')
print(a+1) # Suma de una constante
print('\na*b')
print(a*b) # Cuando dos tensores tienen la misma forma, se suman punto a punto
print('\na-c')
print(a-b) # Cuando dos tensores tienen forma distinta pero compatible, se propaga
print('\nmatmul(a**c, d)')
print(a**c @ d) # El operador @ es la multiplicacion matricial
```

```python
print(a[:2, :3])

a[:2] = b[:2] # Se pueden asignar matrices siempre y cuando sean compatibles
print()
print(a)
print()
print(np.exp(a)) # Funcion exponencial e**a

print()
for row in a:
  print(row)
```

```python
print(a.ravel()) # Aplanado a un array unidimensional
print()
print(a.reshape(2, 25)) # Ahora es una matriz de 2x25
print()
print(a.T) # Hallamos la transpuesta (10x5)
```

```python
from sklearn.datasets import load_wine

data = load_wine()
data
```

### Particionado
Queremos dividir los datos en dos conjuntos, esta es una práctica habitual para ciencia de datos ya que se preserva una muestra de la población sin contaminar para que, en el futuro, poder validar sobre un 'caso real'.

Elegimos los datos, las etiquetas, definimos un 90% de datos para un conjunto y 10% restante para otro, una selección aleatoria proporcional al objetivo mediante *stratify*

```python
from sklearn.model_selection import train_test_split

train_x, test_x, train_y, test_y = train_test_split(data['data'], data['target'],
                                                    train_size=.9, stratify=data['target'])
```

### Estandarizacion
Los modelos suelen ser sensibles a valores con magnitudes diferentes. Vamos a utilizar MinMaxScaler para equiparar las columnas.

Declaramos un MinMaxScaler y con fit_transform encontramos el minimo, maximo y finalmente normalizamos entre 0 y 1. Transform solo ejecuta el ultimo paso y es útil para el conjunto de test preservado.

```python
from sklearn.preprocessing import MinMaxScaler

mms = MinMaxScaler()
train_std = mms.fit_transform(train_x)
test_std = mms.transform(test_x)
```

### Aprendizaje
Vamos a ajustar una máquina de vectores de soporte para clasificar las etiquetas de 'target'. Al igual que el MinMaxScaler, solo hay que hace .fit() para entrenarlo.

```python
from sklearn.svm import SVC

svc = SVC()
svc.fit(train_x, train_y)
```

### Evaluacion
Los modelos suelen contar con una función de score que podemos utilizar. Le indicamos los datos de entrada y sus correspondientes etiquetas para calcular la precisión del modelo como n_aciertos/n_ejemplos

```python
svc.score(test_x, test_y)
```
